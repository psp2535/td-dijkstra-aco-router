# IM-VRM Data Exploration: T-Drive Trajectories
This notebook explores the raw and cleaned Microsoft T-Drive taxi trajectory datasets.

```python
import os
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from src.utils.helpers import load_config
```

```python
# Load configuration
config = load_config()
processed_csv = os.path.join("..", config["data"]["processed_dir"], "clean_trips.csv")

if os.path.exists(processed_csv):
    df = pd.read_csv(processed_csv, parse_dates=["timestamp"])
    print(f"Loaded {len(df)} cleaned points across {df['trip_id'].nunique()} trips.")
    display(df.head())
else:
    print(f"Processed dataset not found at {processed_csv}. Run `python main.py` first.")
```

## Trajectory Statistics
Let's look at velocity, travel distance, and time gap distributions.

```python
if 'df' in locals():
    # Describe speeds
    display(df[['speed_kmh', 'delta_dist_m', 'delta_time_sec', 'acceleration_m_s2']].describe())
    
    # Plot speed histogram
    plt.figure(figsize=(10, 4))
    plt.hist(df['speed_kmh'], bins=50, color='skyblue', edgecolor='black')
    plt.title('Instantaneous Speed Distribution (km/h)')
    plt.xlabel('Speed (km/h)')
    plt.ylabel('Frequency')
    plt.grid(True, alpha=0.3)
    plt.show()
```
